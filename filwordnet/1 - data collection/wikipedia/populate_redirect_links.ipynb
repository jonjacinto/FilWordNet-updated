```python
import os, json
import re
```

```python
# Where you want the populated files will be stored
target_path = '/content/drive/Shared drives/DOST FilWordNet x WordSense/Thesis/Data Collection/Raw Data/Wikipedia/populated/'
```

```python
not_found = []
total_redirects = 0
```

```python
for file_name in [file for file in os.listdir(path) if file.endswith('.json')]:
  with open(path + file_name) as json_file:
    data = json.load(json_file)
    # Extract text
    text = data['full_article']
    # Search for REDIRECTS
    redirect_file = re.search("#REDIRECT \[\[(.*?)\]\]",text)
    if redirect_file:
      total_redirects += 1
      # Replace space with dash
      redirect_file = redirect_file.group(1).replace(" ", "-")+".json"
      print(file_name + " FOUND! " + redirect_file)
      # Open redirect file
      try:
        file2 = open(path + redirect_file)
        data2 = json.load(file2)
        # Replace redirect
        data['full_article'] = ""
        data['full_article'] = data2['full_article']
      except FileNotFoundError:
        not_found.append(redirect_file)
        print(redirect_file + " not found!")
    # Transfer to target directory
    with open(target_path + file_name, "w") as fl:
      json.dump(data, fl)
```

```python
print(len(not_found))
print(total_redirects)
print(len(not_found)/total_redirects)
```
