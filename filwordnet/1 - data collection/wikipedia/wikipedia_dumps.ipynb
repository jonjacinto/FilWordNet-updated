```python
import untangle
import json
import os
```

```python
doc = untangle.parse(path_xml)
```

```python
# Where the parsed files will be stored
path = '/content/drive/Shared drives/DOST FilWordNet x WordSense/Thesis/Data Collection/Raw Data/Wikipedia/json/'
```

```python
count = 0
for page in doc.mediawiki.page:   
    # print("~~~TIMESTAMP~~~")
    # print(page.revision.timestamp.cdata)
    date = page.revision.timestamp.cdata
    
    # print("~~~TITLE~~~")
    # print(page.title.cdata)
    title = page.title.cdata

    # print("~~~REVISION ID~~~")
    # print(page.page.id.cdata)
    page_id = page.id.cdata

    author = ""
    if hasattr(page.revision.contributor, 'name'):
      # print("~~~CONTRIBUTOR NAME~~~")
      # print(page.revision.contributor.username.cdata)
      author = page.revision.contributor.username.cdata
    
    author_id = ""
    if hasattr(page.revision.contributor, 'id'):
      # print("~~~CONTRIBUTOR ID~~~")
      # print(page.revision.contributor.id.cdata)
      author_id = page.revision.contributor.id.cdata

    author_ip = ""
    if hasattr(page.revision.contributor, 'ip'):
      # print("~~~CONTRIBUTOR IP~~~")
      # print(page.revision.contributor.ip.cdata)
      author_ip = page.revision.contributor.ip.cdata
    
    # print("~~~REVISION ID~~~")
    # print(page.revision.id.cdata)
    revision_id = page.revision.id.cdata

    # print("~~~REVISION DATA~~~")
    full_article = " "
    for text in page.revision.text:
         full_article += text.cdata
         full_article += " "
    content = {
        "title": title,
        "author": author,
        "author_id": author_id,
        "author_ip": author_ip,
        "date": date,
        "revision_id": author_ip,
        "full_article": full_article,
    }

    file_name = "%s.json" % title.replace(" ","-")
    print(file_name)
    # if not os.path.isdir('/content/drive/Shared drives/DOST FilWordNet x WordSense/Thesis/Data Collection/Wikipedia/json'):
    #   os.makedirs('/content/drive/Shared drives/DOST FilWordNet x WordSense/Thesis/Data Collection/Wikipedia/json')
    # path = '/content/drive/Shared drives/DOST FilWordNet x WordSense/Thesis/Data Collection/Wikipedia/json/
    try:
      with open(path + file_name, "w") as fl:
        json.dump(content, fl)
        count += 1
    except:
      print("Exception occured")
```

```python
print('Updated raw json count: ', len(doc.mediawiki.page), count)
```
