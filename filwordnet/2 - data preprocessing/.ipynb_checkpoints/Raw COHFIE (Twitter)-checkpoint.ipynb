```python
# Imports
import glob
import os
import pandas as pd
import numpy as np
import regex as re
import html
```

# Load Data

```python
base = "<PATH TO YOUR RAW TWITTER DATA"
source = "Twitter"
year = "2022"
month = "2"
path = f"{base}/{source}/{year}/{month}"
%cd {path}
```

```python
# Load all .json

# the path to your csv file directory
data_dir = path

# get all the csv files in that directory (assuming they have the extension .csv)
json_files = glob.glob(os.path.join(data_dir, '*.json'))

# loop through the files and read them in with pandas
dataframes = []  # a list to hold all the individual pandas DataFrames
for file in json_files:
    df = pd.read_json(file, orient='records', lines=True)
    dataframes.append(df)

# concatenate them all together
df = pd.concat(dataframes, ignore_index=True)

# Remove duplicates
df = df.drop_duplicates()

# show dataframe
display(df)
```

# Preprocess

```python
def preprocess(text):
    # html
    result = html.unescape(html.unescape(str(text)))

    # Email
    result = re.sub(r"[\SÑñ]+@([\SÑñ]+\.)+[\SÑñ]+", " XX_EMAIL ", result)

    # urls
    result = re.sub(r"https?:\/\/([\w\-_]+\.)+([\w\-_]+)+(\/[^\s]+)*", " XX_URL ", result)
    result = re.sub(r"([\w\-_]+\.)+(com|net|org|co|us|ph|io)(\/[^\s]+)*", " XX_URL ", result, flags=re.IGNORECASE)
    
    # twitter username mentions
    result = re.sub(r"(?<!\S)@[^\s.,!?]+(?!\S)", " XX_USERNAME ", result)

    # extra spaces
    result = re.sub(r'\s\s+',' ', result) # replace 2++ white space with 1 white space

    # source specific artifacts removal goes here

    
    return result.strip()
```

```python
%time cleaned = [preprocess(text) for text in df['text']]

# preview cleaned text
cleaned[:5]
```

```python
# Sanity check
print(f"dataframe rows: {df.shape[0]}")
print(f"cleaned: {len(cleaned)}")
```

# Save to JSON

```python
# Build the dataframe to be saved to JSON
df_res = df[['text', 'favorite_count', 'retweet_count', 'created_at', 'location']] # select cols to keep

%time df_res['text'] = cleaned # Replace with preprocessed text

display(df_res)
```

```python
# Save to json
source_type = "social_media"
source = "twitter"
COHFIE_PATH = f"<CHANGE_PATH_HERE>/RAW_COHFIE/{source_type}/{source}/{year}/{month}"

df_res.to_json(path_or_buf=f'{COHFIE_PATH}/{month}.json', orient="records")  
```
