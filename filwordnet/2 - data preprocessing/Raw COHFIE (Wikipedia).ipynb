```python
def clean_wikipedia(text):
  result = str(text)
  result = re.sub(r"<ref>.*</ref>","",result)
  result = re.sub(r"[A-z]+=\".+\"", "", result)
  result = re.sub(r"<small>.*</small>","",result)
  result = re.sub(r"<.+?>", "", result)
  result = re.sub(r"{{.*}}", "", result)
  result = re.sub(r"{{[\S\s]*}}", "", result)
  result = re.sub(r"\[.*\]", "", result)
  result = re.sub(r"\(.*\)", "", result)
  result = re.sub(r"'+","",result)
  result = re.sub(r"=|\*|\(|\)","",result)
  result = re.sub(r"^#REDIRECT(\s)*","",result)
  # result = re.sub(r"\*","",result)
  return result
```

```python
def clean_text_general(text):
    result = str(text)
    result = re.sub(r"([^\s\.,?!]*(\d+,+)*(\d+\.+)?\d+[^\s\.,?!]*)", "", result)
    result = re.sub(r"(\S*_\S*)+", "", result)
    result = re.sub(r"#\S+", "", result)
    result = re.sub(r"https?:\/\/\S*", "", result)
    result = re.sub(r"(\w*\.)+(com|net|org|co|us|ph)(\/\w*)*", "", result)
    result = re.sub(r"-(?!\w)|(?<!\w)-", "", result)
    result = re.sub(r"([.]+)+", ".", result)
    result = re.sub(r"\s+\.\s+", ". ", result)
    result = re.sub(r"\s\s+", " ", result)
    result = re.sub(r"(\[(.*?)\])|(\{(.*?)\})|(\((.*?)\))", "", result)   
    result = re.sub(r"[^a-zA-Z0-9\-\s\.!,?\[\]\(\)Ññ]", "", result)
    result = re.sub(r"\((.*?)\)", "", result)
    result = re.sub(r"\[(.*?)\]", "", result)
    result = re.sub(r"", "", result)
    result = re.sub(r"", "", result)

    return result.strip()
```
