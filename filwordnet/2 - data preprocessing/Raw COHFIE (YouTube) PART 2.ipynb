```python
import json
import pandas as pd
import os
from os.path import exists
import glob
import datetime
import gc
from multiprocessing import Pool
import time
import numpy as np
```

```python
DATA_DIR = r"CHANGE_PATH_HERE"
SAVE_DIR = r"CHANGE_PATH_HERE"
```

```python
csv_files = glob.glob(os.path.join(DATA_DIR, 'PID_2022*.csv'))
csv_files
```

```python
df = pd.read_csv(csv_files[0], index_col=0)
df
```

```python
dfs = []
for csv in csv_files:
    df = pd.read_csv(csv, index_col=0)
    dfs.append(df)
    
    df_all = pd.concat(dfs)
```

```python
df_all.shape
```

```python
years = np.sort(df_all['year'].unique())

for year in years:
    filepath_year = fr"{SAVE_DIR}\{year}"
    try:
        os.mkdir(filepath_year)
    except:
        print(f"{filepath_year} already exists...")
        
    df_year = df_all.loc[df_all['year'] == year]
    
    months = np.sort(df_year['month'].unique())
    
    for month in months:
        filepath_month = fr"{filepath_year}\{month}"
        try:
            os.mkdir(filepath_month)
        except:
            print(f"{filepath_month} already exists...")
            
        df_year.loc[df_year['month'] == month].to_json(f"{filepath_month}\{month}.json", orient="records")
```
