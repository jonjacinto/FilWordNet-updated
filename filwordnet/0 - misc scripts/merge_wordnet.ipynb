```python
import torch

def text_to_embs(sentence_list_texts):
    sentence_embeddings = model.encode(sentence_list_texts, show_progress_bar=False, normalize_embeddings=True)

    return torch.tensor(sentence_embeddings)

def mean_embs(sentence_embs):
    sense_embeddings = torch.mean(sentence_embs, dim=0)

    return sense_embeddings

def get_sentence_embs_df(df):
    df['sense_embeddings'] = None

    for idx, row in df.iterrows():
        # get example sentences
        sentence_list_texts = row['example_sentences']

        # encode to sentence embeddings
        sentence_embeddings = text_to_embs(sentence_list_texts)

        # mean sentence embeddings
        sense_embeddings = mean_embs(sentence_embeddings)

        # append sense embeddings to dataframe row
        row['sense_embeddings'] = sense_embeddings

    return df
```

```python
import torch
import torch.nn.functional as F

def merge_wn(nlp_wn, ns_wn, thres):
    new_wn = pd.DataFrame()
    new_wn['merge_label'] = None

    # Initialize 
    cos = torch.nn.CosineSimilarity(dim=1, eps=1e-08)

    # Get vocab intersection
    nlp_vocab = set(nlp_wn['word'].tolist())
    ns_vocab = set(ns_wn['word'].tolist())
    vocab_combined = nlp_vocab.intersection(ns_vocab)  

    for w in vocab_combined:
        ns_rows = ns_wn.loc[ns_wn['word'] == w]
        nlp_rows = nlp_wn.loc[nlp_wn['word'] == w]

        try:
            ns_se = torch.stack(ns_rows['sense_embeddings'].tolist())
        except:
            print(w)
            print(len(ns_rows['sense_embeddings'].tolist()))
            print(ns_rows['sense_embeddings'].tolist())
            print([t.shape for t in ns_rows['sense_embeddings'].tolist()])

        nlp_se = torch.stack(nlp_rows['sense_embeddings'].tolist())
        
        # cosine similarity = normalize the vectors & multiply
        similarity_matrix = F.normalize(nlp_se) @ F.normalize(ns_se).t()

        max_values, indices = torch.max(similarity_matrix, dim=1)


        # apply threshold
        for idx, mv in enumerate(max_values):
            if mv < thres: # below threshold, append row
                indices[idx] = -1 # sentinel value, marks the rows that will be ignored later

        for idx_nlp, idx_ns in enumerate(indices):
            idx_ns_scalar = idx_ns.item()
            
            if idx_ns_scalar != -1:
                row_nlp = nlp_rows.iloc[idx_nlp].append(pd.Series({'merge_label': idx_ns_scalar}))
                new_wn = new_wn.append(row_nlp, ignore_index=True)
                row_ns = ns_rows.iloc[idx_ns_scalar].append(pd.Series({'merge_label': idx_ns_scalar}))
                new_wn = new_wn.append(row_ns, ignore_index=True)
            else:
                row_nlp = nlp_rows.iloc[idx_nlp].append(pd.Series({'merge_label': -1})) # -1 means no group
                new_wn = new_wn.append(row_nlp, ignore_index=True)
        
        sense_ids = new_wn['sense_id'].tolist()
        ns_sense_ids = [id for id in sense_ids if "ns" in id]

        for idx, row_ns in ns_rows.iterrows():
            if row_ns['sense_id'] not in ns_sense_ids:
                row_ns = row_ns.append(pd.Series({'merge_label': -1})) # -1 means no group
                new_wn = new_wn.append(row_ns, ignore_index=True)


    # drop duplicates
    #new_wn = new_wn.drop_duplicates()
    new_wn = new_wn.drop_duplicates(subset=['sense_id']).reset_index(drop=True)

    return new_wn
```

```python
def aggregate_contextual_info(contextual_info_lst):
    res = {}
    for i in contextual_info_lst:
        for k_sourcetype in i:
            v_sourcetype = i[k_sourcetype]
            # print(k_sourcetype)
            # print(f"    {v_sourcetype}")

            if k_sourcetype not in res: # first time
                res[k_sourcetype] = {}

            for k_source in v_sourcetype:
                v_source = v_sourcetype[k_source]
                # print(f"        {k_source}")
                # print(f"        {v_source}")

                if k_source not in res[k_sourcetype]: # first time
                    res[k_sourcetype][k_source] = {}

                for k_year in v_source:
                    v_year = v_source[k_year]
                    # print(f"            {k_year}")
                    # print(f"            {v_year}")

                    if k_year not in res[k_sourcetype][k_source]: # first time
                        res[k_sourcetype][k_source][k_year] = v_year
                    else:
                        res[k_sourcetype][k_source][k_year] += v_year
                        
    return res
```
