# Generate Synsets
This notebook contains the code for Synset Induction Module

This takes the sense inventory as an input and outputs the wordnet i.e. sense inventory + synset id

```python
import pandas as pd
from sklearn.cluster import AgglomerativeClustering
```

```python
def generate_synsets(sense_inventory):
    '''
        df_sense_embeddings: columns=['word', 'sense_id', 'example_sentences', 'pos', 'contextual_info', 'sense_embedding']
    '''
    df_synsets = sense_inventory.copy()
    sense_embeddings = list(sense_inventory['sense_embedding'])
    
    distance_threshold = 0.12
    af = AgglomerativeClustering(n_clusters=None, 
                                     affinity="cosine", 
                                     linkage="average", 
                                     distance_threshold=distance_threshold).fit(sense_embeddings)

    df_synsets['synset_id'] = af.labels_

    return df_synsets
```

```python
sense_inventory = pd.read_pickle(f"sense_inventory.pkl")
sense_inventory
```

```python
wordnet = generate_synsets(sense_inventory)
wordnet
```

## Save wordnet to file

```python
wordnet.to_csv(r"./filwordnet.csv", index=False) # To CSV
wordnet.to_pickle(r"./filwordnet.pkl", protocol=3) # To pickle
```
