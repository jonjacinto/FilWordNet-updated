# Load translate words

```python
df_valid = pd.read_csv('full_vocab_translated.csv')
df_valid
```

## Query WordNet

```python
words = df_valid['eng_text'].values.tolist()
df = get_wordnet_entries(words)
```

```python
df
```

## Translate

```python
res = translate_df(df, mode="en_to_tl", backtranslate=False)
res
```

```python
res.head()
```

## Save 

```python
res.to_csv('full_vocab_evaluation.csv', index=False)
```

```python
res = res.rename(columns={'word': 'eng_word'}).set_index('eng_word')
res
```

```python
# join with df_valid
df_final = df_valid.join(res, on=["eng_word"], how="left").fillna('XX')
df_final
```
