```python
import pandas as pd
import os
import json

import pymongo
from pymongo import MongoClient
from bson.objectid import ObjectId
from bson.json_util import dumps, loads
from bson import json_util
```

```python
temp_path = os.getcwd().split('\\')
project_dir = '\\'.join(temp_path[:temp_path.index('SOURCE') + 1])
```

```python
# TODO: Update the connection string
connection_string = ''

client = pymongo.MongoClient(connection_string)
db = client["WordNet"]
```

#Fill DB

```python
path = f'{project_dir}/data/8 - Wordnet Creation/wordnet_entries'

wordnet = [pos_json for pos_json in os.listdir(path)]
```

```python
for senses in wordnet:
  with open(f'{path}/{senses}') as json_file:
    data = json.load(json_file)
  
  results = db["NS_SenseInventory"].insert_one(data)
  
  print("New sense are added with IDs: ", results)
```

#Find senses in DB

```python
# Finds all senses of a target word
target = "alam" # update the target word
query = db["NS_SenseInventory"].find({'word': target})
print(loads(dumps(query)))
```
