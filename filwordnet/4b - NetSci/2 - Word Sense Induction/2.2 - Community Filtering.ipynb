# Filter Small Communities

```python
algorithm = 'leiden_mod' # TODO: update

old_path = f'{project_dir}/data/4 - Word Sense Induction/communities/{algorithm}'
new_path = f'{project_dir}/data/4 - Word Sense Induction/communities/{algorithm}_filtered_10'
```

```python
if not os.path.isdir(new_path):
    os.makedirs(new_path)
```

```python
filtering_threshold = 0.10 # TODO: update
```

```python
files = glob(f'{old_path}/*_{algorithm}.json')

for filename in files:
    try:
        with open(filename, 'r') as f:
            data = json.load(f)
            old_communities = data['community'].values()
            threshold = int(sum([len(comm) for comm in old_communities]) * filtering_threshold)
            new_communities = {index: comm for index, comm in enumerate(old_communities) if len(comm) > threshold}
            if len(new_communities) < len(old_communities):
                  print(f'{filename.replace(old_path, "")} removed {len(old_communities) - len(new_communities)}')
            data['community'] = new_communities

        with open(filename.replace(old_path, new_path), 'w') as f:
            json.dump(data, f)
    except:
        print(filename)
```
