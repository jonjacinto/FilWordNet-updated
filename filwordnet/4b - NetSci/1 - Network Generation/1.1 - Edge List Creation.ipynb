### Iterative Processing

```python
directory = f"{project_dir}/data/2 - COHFIE Creation/COHFIE"
source_type = 'online_forums' # TODO: change
source = 'reddit' # TODO: change
```

```python
path = f'{directory}/{source_type}/{source}'

# directory iterate from year down to months 
years = sorted(os.listdir(path))
for year in years:
        months = sorted(os.listdir(f'{path}/{year}'))
    
        for month in months:
          file_path = f'{path}/{year}/{month}'
          json_files = sorted(glob.glob(f'{file_path}/*[!_POS].json'))
          pos_files = sorted(glob.glob(f'{file_path}/*_POS.json'))
          print("DATE: ", year, "/", month)
          # loop through the files and read them in with pandas
          dataframes = []  # a list to hold all the individual pandas DataFrames
          for index in range(len(json_files)):
              print(f"Importing {json_files[index].split('/')[-1]}...")
              pos_df = pd.read_json(open(pos_files[index]), orient="records") 
              original_df = pd.read_json(open(json_files[index]), orient="records").drop(['pos_tags', 'text'], axis=1) 

              df = pos_df.join(original_df)

              df = df.loc[(((df['lang'] == 'en') & (df['lang_prob'] < 0.8)) | (df['lang'] == 'fil')) & (df['pos_tags'].map(len) > 4)]

              dataframes.append(df)

          start_time = time.time()
          edgelist = preprocessing.create_edgelist(pd.concat(dataframes, ignore_index=True), window_size)



          print(edgelist.head(5))

          if not edgelist.empty:
              edgelist, nodelist_dict = preprocessing.update_nodelist(edgelist)
              preprocessing.save_edgelist(edgelist)
              preprocessing.save_nodelist(nodelist_dict)
              print("Saved edge list and node list.")

          print("--- %s seconds ---" % (time.time() - start_time))
          print("\n================================\n")
```

### Single File Processing

```python
directory = f"{project_dir}/data/2 - COHFIE Creation/COHFIE"
source_type = 'online_forums' # TODO: change
source = 'reddit' # TODO: change
year = 2021 # TODO: change
month = 10 # TODO: change
```
