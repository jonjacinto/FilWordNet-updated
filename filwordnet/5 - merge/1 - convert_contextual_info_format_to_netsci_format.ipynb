```python
data = pd.read_pickle(NLP_WORDNET_PATH)
data
```

```python
def get_source_type(source):
    if source in ['twitter', 'youtube']:
        return 'social_media'
    elif source in ['abante', 'abscbn', 'balita', 'bandera', 'gma', 'mb', 'philstar','radyoinquirer']:
        return 'news_sites'
    elif source in ['bible', 'google_books', '100year', 'gutenberg']:
        return 'books'
    elif source in ['reddit']:
        return 'online_forums'
    elif source in ['wikipedia']:
        return 'online_encyclopedia'
```

```python
# Initialize list
contextual_info = []

# Convert to NetSci format
for context in data['contextual_info']:
    json_context = context
    contextual_info_entry = {}
    
    for source in json_context:
        source_type = get_source_type(source)

        if source_type not in contextual_info_entry:
            contextual_info_entry[source_type] = {}
            
        for year_str in json_context[source]:

            if source not in contextual_info_entry[source_type]:
                contextual_info_entry[source_type][source] = {}
                
            counts = json_context[source][year_str]
            contextual_info_entry[source_type][source][year_str] = counts
    
    contextual_info.append(contextual_info_entry)

# Update dataframe
data['contextual_info'] = contextual_info
```
