## COHFIE 
❓ The Corpus of Historical Filipino and Philippine English (COHFIE) is a language resource containing sentences from 5 different categories of textual data that was pre-processed by the team for other researchers to use.

The following is the JSON response model of all COHFIE endpoints
- text: String containing a sentence from a document
-  pos_tags: String containing the POS tags of each word in the text
- source: String containing the specific source from where the text was collected
- source_type: String containing the general category of the source
- year: Integer containing the year when the text was made
- month: Integer containing the month when the text was made

### **URL:** `/get_sentences`
Retrieves sentences based on a target query word, and/or other filtering parameters.

The parameters are as follows:
- **query**: String that must be present in all sentences (required) 
- **start_year**: Integer that indicates the start year of the sentences. Default set to `0`.
- **end_year**: Integer that indicates the end year of the sentences. Default set to `999999`.
- **start_month**: Integer that indicates the start month of the sentences. Default set to `0`.
- **end_month**: Integer that indicates the end month of the sentences. Default set to `999999`.
- **source_type**: String that indicates the source of the sentences. Default set to `None`.
  - Accepted values are: online_forums, online_encyclopedia, news_sites, social_media, books.
- **limit**: Integer that indicates the maximum number of sentences to be returned. Default set to `0`.

To further demonstrate the usage of each parameter, here is an example of using the endpoint to retrieve all sentences that contain the word "*pinto*". It is important to keep in mind that the **query** parameter is always required when calling the API.

````python
# Sample request through Python
import requests
res = requests.get(PUBLIC_URL+"/get_sentences?query=pinto")
# Print the first 2 results
res.json()[:2]
````

## WordNet

The WordNet contains the diachronic representations of Filipino words and senses through a community of words and sense embeddings that were generated by our models using COHFIE data. Each entry contains the word, the sense, the sense representation, the synset, example sentences containing the sense of the word, and the sense representation itself. <br/><br/>

The following is the JSON response model of the WordNet endpoint:
- **sense_id**: String containing the sense id 
- **word**: String containing the word label
- **synset_id**: String containing the synset id
- **example_sentences**: List of strings (sentences) containing the sense of the word
- **contextual_info**: Object containing the contextual info of the sense representation
<br/><br/>

Another field in the response is the sense representation of the sense of a word which differs according to the endpoint used.
<br/><br/>

For the `/get_netsci_word/` endpoint:
- **community**: List of strings (words) that belong to a single community after community detection of the network
<br/><br/>

For the `/get_word` endpoint:
- **sense_embeddings**: Vector generated by the BERT model that represents a specific sense of the word 


### **URL:** `/get_netsci_word/`
Returns the wordnet entry of the word, its sense/s, and their sense representation that were generated by using a network. (Banogon, et. al)
<br/>
The parameters are as follows:
- **word**: String containing the target word to be queried (required)
- **show_context**: Boolean indicating whether the contextual info should be shown. Default set to `false`.

````python
# Sample request through Python
import requests
res = requests.get(PUBLIC_URL + "/get_netsci_word/?word=pera")
# Print the first result
res.json()[:1]
````

Below is the output of the code shown above.

```json
{
  'community': ['lagi','lagyan','resibo',...,pangbayad],
  'example_sentences': [tapos nanghihiram na sila ng pera sakin which is nakakagulat ... sobrang busy ko talaga .'
```

In order to show the context, set the **show_context** parameter to true.

## Gold Standard
A gold standard was made by Banogon et al. in order to compare the WSD results of their model against a ground truth. The sentences in this collection are already pre-processed and are utilized in the validation module of Banogon et al.<br/><br/>

The following is the JSON response model of the Gold Standard endpoint:
- **label**: String containing the word label
- **numUsers**: Integer containing the number of annotators for that word
- **sentences**: List of strings (sentences) containing the word  

### **URL** `/get_gs_sentences`
Retrieves all the gold standard sentences
